<h2 style='color:#0891B2' align='center'>Working with OpenAI api</h2>

OpenAi api allows us to access and customise their models. API stands for Application Programming interface ad acts as messenger between software applications.

```python
from openai import OpenAI
```

```python
client = OpenAI(
    api_key=""
    )
```

We can choose the role as then add the question we want to ask in the messages like in the code below

```python
response = client.chat.completions.create(
                                            model="gpt-4o-mini",
                                            messages=[{
                                                        "role": "user",
                                                        "content": "What is the OpenAI API?"
                                                        }]
                                        )
```

```python
print(response.choices[0].message.content)
```

<h4 style='color:#0891B2' align='left'>Ask openai api to give you 5 productivity tips</h4>

```python
response = client.chat.completions.create(
  model="gpt-4o-mini",
  messages=[{"role": "user", "content": "Give me 5 Quick productivity tips in numbered list with one sentence each."}]
)

# Extract the content from the response
print(response.choices[0].message.content)
```

<h4 style='color:#0891B2' align='left'>Find and replace feature</h4>

We can use max_completion_tokens to control the maxiumum length of response from api

```python
prompt="""
Replace car with plane and adjust phrase:
A car is a vehicle that is typically powered by an internal combustion engine or an electric motor. 
It has four wheels, and is designed to carry passengers and/or cargo on roads or highways. 
Cars have become a ubiquitous part of modern society, and are used for a wide variety of purposes, 
such as commuting, travel, and transportation of goods. Cars are often associated with freedom, 
independence, and mobility."""

# Create a request to the Chat Completions endpoint
response = client.chat.completions.create(
  model="gpt-4o-mini",
  messages=[{"role": "user", "content": prompt}],
  max_completion_tokens=100
)

# Extract and print the response text
print(response.choices[0].message.content)
```
