<h2 style='color:#0891B2' align='center'>Working with OpenAI api</h2>

OpenAi api allows us to access and customise their models. API stands for Application Programming interface ad acts as messenger between software applications.

```python
from openai import OpenAI
```

```python
client = OpenAI(
    api_key=""
    )
```

We can choose the role as then add the question we want to ask in the messages like in the code below

```python
response = client.chat.completions.create(
                                            model="gpt-4o-mini",
                                            messages=[{
                                                        "role": "user",
                                                        "content": "What is the OpenAI API?"
                                                        }]
                                        )
```

```python
print(response.choices[0].message.content)
```
